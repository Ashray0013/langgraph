```python
from typing import List,  TypedDict 
from langgraph.graph import StateGraph
```

```python
class InputData(TypedDict):
    name: str
    operation: str
    values: List[int]
    result: str
```

```python
def processInput(state: InputData) -> InputData:
    if state["operation"] == "+":
        state["result"] = str(sum(state["values"]))
        state["result"] = f"Hi there {state['name']}, the sum of your values is {state['result']}."
    elif state["operation"] == "*":
        product = 1
        for value in state["values"]:
            product *= value
        state["result"] = str(product)
        state["result"] = f"Hi there {state['name']}, the product of your values is {state['result']}."
    else:
        state["result"] = "Unsupported operation"

    
    return state
```

```python
graph = StateGraph(InputData)

graph.add_node("processor", processInput)
graph.set_entry_point("processor")
graph.set_finish_point("processor")

app = graph.compile()
```

```python
result = app.invoke({"name": "Ashray","operation": "+", "values": [1, 2, 3], "result": ""})
print(result["result"])  # Output: 'Hi there Ashray, the sum of your values is 6'
result = app.invoke({"name": "Ashray","operation": "*", "values": [1, 2, 3, 4], "result": ""})
print(result["result"])  # Output: 'Hi there Ashray, the product of your values is 24'
```
